# VulHunter — Binary Vulnerability Detection (Fusion Mode)

Notebook này huấn luyện **VulHunterModel** ở chế độ **fusion** — kết hợp:
- **Semantic Encoder**: `Qwen2.5-Coder-1.5B-Instruct` với LoRA (r=16, α=32)
- **Graph Encoder**: `GraphCodeBERT` (unfreeze top-6 layers) + GAT
- **Cross-Modal Fusion**: Gated bidirectional cross-attention với residual skip
- **Binary Head**: Predict vulnerable (1) vs safe (0)

## 3. Khám phá dataset (EDA)

```python
def quick_eda(path: Path, n: int = 3000):
    rows = []
    with path.open() as f:
        for i, line in enumerate(f):
            if i >= n: break
            r = json.loads(line)
            rows.append({
                "binary_label": r.get("binary_label"),
                "quality_tier": r.get("quality_tier", "gold"),
                "code_len": len(r.get("code", "")),
                "has_tokens": "input_ids_qwen" in r,
            })
    df = pd.DataFrame(rows)
    print(f"Loaded {len(df)} samples")
    print(f"\nBinary label:")
    print(df["binary_label"].value_counts(normalize=True).round(3))
    print(f"\nQuality tier:")
    print(df["quality_tier"].value_counts(normalize=True).round(3))
    print(f"\nCode length:")
    print(df["code_len"].describe().round(0))
    return df

eda_df = quick_eda(TRAIN_JSONL, n=3000)
```

## 4. Cấu hình training (Fusion mode)

```python
MODE = "fusion"
MAX_LEN = 2048
BATCH_SIZE = 2
GRAD_ACCUM = 4
EPOCHS = 8
LR_BACKBONE = 2.0e-5
LR_GRAPH = 1.0e-4
LR_HEAD = 2.0e-4
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1
FOCAL_ALPHA = AUTO_FOCAL_ALPHA
FOCAL_GAMMA = 2.0
PATIENCE = 2
NUM_WORKERS = 2
USE_AMP = True
TUNE_THRESHOLD = True  # Auto tune threshold trên val set

SEMANTIC_CFG = {
    "backbone": "Qwen/Qwen2.5-Coder-1.5B-Instruct",
    "output_dim": 256,
    "freeze_layers": 28,
    "dropout": 0.1,
    "pooling": "last",
    "gradient_checkpointing": True,
    "use_fp16": True,
    "use_lora": True,
    "lora_r": 16,
    "lora_alpha": 32,
    "lora_dropout": 0.05,
    "use_rslora": True,
    "lora_target_modules": ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
}

GRAPH_CFG = {
    "use_graphcodebert": True,
    "node_feature_dim": 128,
    "hidden_dim": 256,
    "output_dim": 256,
    "num_layers": 4,
    "num_heads": 8,
    "num_edge_types": 5,
    "dropout": 0.2,
    "unfreeze_top_n": 6,  # ★ chống AUC=0.5 collapse
}

FUSION_CFG = {
    "hidden_dim": 256,
    "num_heads": 8,
    "num_layers": 2,
    "dropout": 0.1,
    "combine": "gated",
    "residual_alpha": 0.3,  # ★ giữ semantic signal
}

HEAD_CFG = {
    "binary": {"hidden_dim": 128, "dropout": 0.3},
}

print(f"Mode: {MODE}")
print(f"Semantic: LoRA r={SEMANTIC_CFG['lora_r']}/α={SEMANTIC_CFG['lora_alpha']}, pooling={SEMANTIC_CFG['pooling']}")
print(f"Graph: unfreeze_top_n={GRAPH_CFG['unfreeze_top_n']}")
print(f"Fusion: combine={FUSION_CFG['combine']}, residual_alpha={FUSION_CFG['residual_alpha']}")
```

## 5. Build model + load datasets

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

model = VulHunterModel(
    mode=MODE,
    semantic_config=SEMANTIC_CFG,
    graph_config=GRAPH_CFG,
    fusion_config=FUSION_CFG,
    head_config=HEAD_CFG,
)
model.to(device)

# Trainable params → float32 cho AMP GradScaler
for p in model.parameters():
    if p.requires_grad and p.dtype != torch.float32:
        p.data = p.data.float()

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"✓ Model [{MODE}] — Trainable: {trainable:,} / {total:,} ({100*trainable/total:.2f}%)")
```

```python
# Build dataloaders
train_ds = VulHunterDataset(data_path=TRAIN_JSONL, max_length=MAX_LEN, graph_data_path=GRAPH_PATH)
val_ds = VulHunterDataset(data_path=VAL_JSONL, max_length=MAX_LEN, graph_data_path=GRAPH_PATH)
test_ds = VulHunterDataset(data_path=TEST_JSONL, max_length=MAX_LEN, graph_data_path=GRAPH_PATH)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate_fn, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())

print(f"Train: {len(train_ds):,} | Val: {len(val_ds):,} | Test: {len(test_ds):,}")

# Sample
sample = next(iter(train_loader))
print("\nSample keys:", list(sample.keys()))
print(f"  input_ids: {sample['input_ids'].shape}")
print(f"  binary_labels: {sample['binary_labels'].tolist()[:8]}")
if "batch" in sample:
    print(f"  graph_batch: {sample['batch'].shape}, nodes={len(sample['node_types'])}")
```

```python
# Sanity check forward pass
model.eval()
sample = next(iter(val_loader))
with torch.no_grad():
    kwargs = {
        "input_ids": sample["input_ids"].to(device),
        "attention_mask": sample["attention_mask"].to(device),
        "node_types": sample["node_types"],
        "node_texts": sample.get("node_texts"),
        "edge_index": sample["edge_index"].to(device),
        "edge_type": sample["edge_type"].to(device),
        "batch": sample["batch"].to(device),
    }
    out = model(**kwargs)
print(f"binary_logits: {out.binary_logits.shape}")
print(f"fused_embedding: {out.fused_embedding.shape}")
print("✓ Forward pass OK")
model.train()
```

## 6. Setup optimizer + criterion + scheduler

```python
# Criterion: Binary focal loss
criterion = BinaryClassificationLoss(focal_alpha=FOCAL_ALPHA, focal_gamma=FOCAL_GAMMA)

# Tiered LR: backbone, graph, head
backbone_params, graph_params, head_params = [], [], []
for name, p in model.named_parameters():
    if not p.requires_grad: continue
    if "semantic_encoder.backbone" in name:
        backbone_params.append(p)
    elif "graph_encoder" in name:
        graph_params.append(p)
    else:
        head_params.append(p)

param_groups = []
if backbone_params: param_groups.append({"params": backbone_params, "lr": LR_BACKBONE})
if graph_params:    param_groups.append({"params": graph_params, "lr": LR_GRAPH})
if head_params:     param_groups.append({"params": head_params, "lr": LR_HEAD})
optimizer = torch.optim.AdamW(param_groups, weight_decay=WEIGHT_DECAY)
print(f"Optimizer: {len(backbone_params)} backbone ({LR_BACKBONE}) + {len(graph_params)} graph ({LR_GRAPH}) + {len(head_params)} head ({LR_HEAD})")

# Cosine + warmup
steps_per_epoch = math.ceil(len(train_loader) / GRAD_ACCUM)
total_steps = steps_per_epoch * EPOCHS
warmup_steps = int(total_steps * WARMUP_RATIO)

def lr_lambda(step):
    if step < warmup_steps:
        return step / max(1, warmup_steps)
    progress = (step - warmup_steps) / max(1, total_steps - warmup_steps)
    return max(0.0, 0.5 * (1.0 + math.cos(math.pi * min(1.0, progress))))

scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda=lr_lambda)
scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP and torch.cuda.is_available())
print(f"Total steps: {total_steps} | Warmup: {warmup_steps}")
```

## 7. Training loop

```python
def train_one_epoch(model, loader, optimizer, scheduler, scaler, epoch):
    model.train()
    total_loss = 0.0
    n = 0
    optimizer.zero_grad(set_to_none=True)
    t0 = time.time()
    for step, batch in enumerate(loader):
        kwargs = {
            "input_ids": batch["input_ids"].to(device, non_blocking=True),
            "attention_mask": batch["attention_mask"].to(device, non_blocking=True),
            "node_types": batch["node_types"],
            "node_texts": batch.get("node_texts"),
            "edge_index": batch["edge_index"].to(device, non_blocking=True),
            "edge_type": batch["edge_type"].to(device, non_blocking=True),
            "batch": batch["batch"].to(device, non_blocking=True),
        }
        is_accumulating = ((step + 1) % GRAD_ACCUM != 0) and ((step + 1) != len(loader))
        if USE_AMP and torch.cuda.is_available():
            with torch.autocast(device_type="cuda", dtype=torch.float16):
                out = model(**kwargs)
                losses = criterion(
                    binary_logits=out.binary_logits,
                    binary_targets=batch["binary_labels"].to(device),
                    sample_weights=batch["sample_weights"].to(device),
                )
                loss = losses["total"] / GRAD_ACCUM
            scaler.scale(loss).backward()
        else:
            out = model(**kwargs)
            losses = criterion(
                binary_logits=out.binary_logits,
                binary_targets=batch["binary_labels"].to(device),
                sample_weights=batch["sample_weights"].to(device),
            )
            loss = losses["total"] / GRAD_ACCUM
            loss.backward()
        if not is_accumulating:
            if USE_AMP and scaler is not None:
                try: scaler.unscale_(optimizer)
                except ValueError:
                    for p in model.parameters():
                        if p.requires_grad and p.dtype != torch.float32:
                            p.data = p.data.float()
                    try: scaler.unscale_(optimizer)
                    except: pass
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer); scaler.update()
            else:
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)
        total_loss += loss.item() * GRAD_ACCUM
        n += 1
        if (step + 1) % 50 == 0:
            ips = (step + 1) / (time.time() - t0)
            lr = optimizer.param_groups[0]["lr"]
            print(f"  E{epoch+1} S{step+1}/{len(loader)} | loss={total_loss/n:.4f} | lr={lr:.2e} | {ips:.2f} it/s")
    return total_loss / max(n, 1)

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    y_true, y_prob = [], []
    for batch in loader:
        kwargs = {
            "input_ids": batch["input_ids"].to(device),
            "attention_mask": batch["attention_mask"].to(device),
            "node_types": batch["node_types"],
            "node_texts": batch.get("node_texts"),
            "edge_index": batch["edge_index"].to(device),
            "edge_type": batch["edge_type"].to(device),
            "batch": batch["batch"].to(device),
        }
        if USE_AMP and torch.cuda.is_available():
            with torch.autocast(device_type="cuda", dtype=torch.float16):
                out = model(**kwargs)
        else:
            out = model(**kwargs)
        p = torch.sigmoid(out.binary_logits.squeeze(-1))
        y_true.extend(batch["binary_labels"].tolist())
        y_prob.extend(p.cpu().tolist())
    return np.array(y_true), np.array(y_prob)

print("✓ Training functions defined")
```

```python
# Main training loop
best_f1 = 0.0
best_threshold = 0.5
patience_counter = 0
history = []
CKPT_DIR = Path("/kaggle/working/models/checkpoints/fusion") if IS_KAGGLE else REPO_DIR / "models/checkpoints/fusion"
CKPT_DIR.mkdir(parents=True, exist_ok=True)

for epoch in range(EPOCHS):
    t0 = time.time()
    train_loss = train_one_epoch(model, train_loader, optimizer, scheduler, scaler, epoch)
    val_y, val_prob = evaluate(model, val_loader)
    elapsed = time.time() - t0

    # Tune threshold
    if TUNE_THRESHOLD and len(val_y) > 0:
        thr, m = find_best_threshold(val_y, val_prob)
    else:
        thr = 0.5
        preds = (val_prob >= thr).astype(int)
        m = binary_metrics(val_y, preds, val_prob, threshold=thr)
    val_f1 = m.f1

    print(f"\nEpoch {epoch+1}/{EPOCHS} ({elapsed:.1f}s)")
    print(f"  Train loss: {train_loss:.4f}")
    print(f"  Val:   F1={val_f1:.4f} | P={m.precision:.4f} | R={m.recall:.4f} | AUC={m.auc:.4f} | thr={thr:.2f}")

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "val_f1": val_f1,
        "val_metrics": m.to_dict(),
        "threshold": thr,
        "elapsed": elapsed,
    })

    if val_f1 > best_f1:
        best_f1 = val_f1
        best_threshold = thr
        patience_counter = 0
        trainable_names = {n for n, p in model.named_parameters() if p.requires_grad}
        sd = {k: v.cpu() for k, v in model.state_dict().items() if k in trainable_names}
        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": sd,
            "best_f1": best_f1,
            "best_threshold": best_threshold,
            "config": {"mode": MODE, "model": {
                "semantic": SEMANTIC_CFG, "graph": GRAPH_CFG,
                "fusion": FUSION_CFG, "heads": HEAD_CFG,
            }},
            "history": history,
        }, CKPT_DIR / "best.pt")
        print(f"  ★ Saved best (F1={best_f1:.4f}, thr={best_threshold:.2f})")
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print(f"\n⏹ Early stopping sau {PATIENCE} epochs không cải thiện.")
            break
    print()

print(f"\n✓ Best Val F1 = {best_f1:.4f} @ threshold {best_threshold:.2f}")
print(f"  Checkpoint: {CKPT_DIR}/best.pt")
```

## 8. Đánh giá trên test set + diagnostic plots

```python
test_y, test_prob = evaluate(model, test_loader)

# Apply best threshold
test_pred = (test_prob >= best_threshold).astype(int)
test_metrics = binary_metrics(test_y, test_pred, test_prob, threshold=best_threshold)

print("=" * 60)
print("BINARY CLASSIFICATION — TEST RESULTS")
print("=" * 60)
for k, v in test_metrics.to_dict().items():
    print(f"  {k}: {v}")
```

```python
# Diagnostic plots
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# 1. Confusion Matrix
cm = confusion_matrix(test_y, test_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=axes[0, 0],
            xticklabels=["Safe", "Vuln"], yticklabels=["Safe", "Vuln"])
axes[0, 0].set_title(f"Confusion Matrix (thr={best_threshold:.2f})")
axes[0, 0].set_xlabel("Predicted")
axes[0, 0].set_ylabel("True")

# 2. ROC Curve
if len(np.unique(test_y)) > 1:
    fpr, tpr, _ = roc_curve(test_y, test_prob)
    auc_val = roc_auc_score(test_y, test_prob)
    axes[0, 1].plot(fpr, tpr, label=f"AUC = {auc_val:.4f}", color="#2ecc71", lw=2)
    axes[0, 1].plot([0, 1], [0, 1], "k--", alpha=0.3)
    axes[0, 1].set_xlabel("FPR")
    axes[0, 1].set_ylabel("TPR")
    axes[0, 1].set_title("ROC Curve")
    axes[0, 1].legend()

# 3. Precision-Recall Curve
p, r, _ = precision_recall_curve(test_y, test_prob)
ap_val = average_precision_score(test_y, test_prob)
    axes[1, 0].plot(r, p, label=f"AP = {ap_val:.4f}", color="#e74c3c", lw=2)
axes[1, 0].set_xlabel("Recall")
axes[1, 0].set_ylabel("Precision")
axes[1, 0].set_title("Precision-Recall Curve")
axes[1, 0].legend()

# 4. Probability distribution
axes[1, 1].hist(test_prob[test_y == 0], bins=50, alpha=0.5, label="Safe", color="#2ecc71")
axes[1, 1].hist(test_prob[test_y == 1], bins=50, alpha=0.5, label="Vuln", color="#e74c3c")
axes[1, 1].axvline(best_threshold, color="k", linestyle="--", label=f"thr={best_threshold:.2f}")
axes[1, 1].set_xlabel("Predicted P(vuln)")
axes[1, 1].set_ylabel("Count")
axes[1, 1].set_title("Probability Distribution")
axes[1, 1].legend()
plt.tight_layout()
plt.savefig(CKPT_DIR / "diagnostics.png", dpi=120)
plt.show()
print(f"✓ Saved diagnostics to {CKPT_DIR}/diagnostics.png")
```
